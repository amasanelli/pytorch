## Matrix multiplication

```python
tensor.matmul(tensor), 1*1+2*2+3*3, tensor @ tensor
```

```python
%%time
tensor.matmul(tensor)
```

```python
%%time
1*1+2*2+3*3
```

```python
%%time
value = 0
for i in rng(len(tensor)):
  value += tensor[i] * tensor[i]
value
```

```python
# Matix multiplication
#  Inner dimensions must match
#  The resulting matrix has the shape of the outer dimensions
torch.rand(3,2) @ torch.rand(2,4)
```

```python
# Transpose
torch.rand(3,2).mm(torch.rand(3,2).T)
```

# Numpy

```python
import numpy as np
```

```python
array = np.arange(1,10)
tensor = torch.from_numpy(array)
array, tensor, tensor.dtype
```

```python
array = np.arange(1.0,10.0)
tensor = torch.from_numpy(array)
array, tensor
```

```python
array += 1
array, tensor
```

```python
array = array + 1
array, tensor
```

```python
tensor = torch.ones(10)
array = tensor.numpy()
tensor, array
```

```python
tensor += 1
tensor, array
```

```python
tensor = tensor + 1
tensor, array
```
